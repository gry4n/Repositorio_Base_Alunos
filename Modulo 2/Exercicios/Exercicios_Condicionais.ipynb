Exercício 5 - Classificação de Notas 🧑‍🎓

Pergunte a nota de um aluno e dependendo da nota que ele informar, o classifique de acordo com as regras abaixo:

Classifique a nota do aluno:

- 9 até 10: "Excelente"

- 7 até 8.9: "Bom"

- 5 até 6.9: "Regular"

- 0 até 4.9: "Reprovado"

mostre para o aluno a classificação dele através de um print, exemplo print("Excelente").

```python
nota = float(input("Qual sua nota?: "))

if nota >= 9.0 and nota <= 10.0:
    print("Excelente")
elif nota >= 7.0 and nota < 9.0:
    print("Bom")
elif nota >= 5.0 and nota < 7.0:
    print("Regular")
elif nota >= 0.0 and nota < 5.0:
    print("Reprovado")
```

Exercício 9 - Calculadora Simples 📏📐

- Faça uma calculadora que:

- Pede dois números

- Pede a operação (+, -, *, /)

- Dependendo da operação mostre o resultado.



```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
operacao = input("Digite a operação desejada (+, -, *, /): ")

if operacao == "+":
    resultado = num1 + num2
    print(f"Resultado: {num1} + {num2} = {resultado}")

elif operacao == "-":
    resultado = num1 - num2
    print(f"Resultado: {num1} - {num2} = {resultado}")

elif operacao == "*":
    resultado = num1 * num2
    print(f"Resultado: {num1} * {num2} = {resultado}")

elif operacao == "/":
    if num2 != 0:
        resultado = num1 / num2
        print(f"Resultado: {num1} / {num2} = {resultado}")
    else:
        print("Erro: Não é possível dividir um número por zero!")

else:
    print("Operação inválida! Por favor, escolha entre +, -, * ou /.")
```
